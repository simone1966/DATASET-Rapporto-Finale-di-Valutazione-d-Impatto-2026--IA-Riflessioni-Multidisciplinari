```python
import os
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = 'data'
df_t0 = pd.read_csv(os.path.join(DATA_DIR, '01_survey_T0_baseline.csv'))
df_t2 = pd.read_csv(os.path.join(DATA_DIR, '02_survey_T2_post.csv'))
df_sr = pd.read_csv(os.path.join(DATA_DIR, '03_second_ring_surveys.csv'))
df_ref = pd.read_csv(os.path.join(DATA_DIR, '04_lab_reflections_corpus.csv'))
df_t1 = pd.read_csv(os.path.join(DATA_DIR, '05_lab_T1_assessments_meta.csv'))

print('1. T0 Baseline rows:', len(df_t0), '| Unique items:', df_t0['item_code'].nunique())
print('2. T2 Post rows:', len(df_t2), '| Unique items:', df_t2['item_code'].nunique())
print('3. Second Ring individual records:', len(df_sr))
print('4. Anonymized Reflections corpus:', len(df_ref))
print('5. T1 Lab assessments metadata:', len(df_t1))
```

```python
# Exploratory Summary 1: Second Ring Cross-Linguistic Cohorts (es-AR, ro, ru)
print('Cohort distribution:')
print(df_sr['cohort_code'].value_counts())

age_col = [c for c in df_sr.columns if c.startswith('A01.')][0]
df_sr[age_col] = pd.to_numeric(df_sr[age_col], errors='coerce')
print('\nAge summary by cohort:')
display(df_sr.groupby('cohort_code')[age_col].agg(['count', 'mean', 'median', 'min', 'max']))
```

```python
# Exploratory Summary 2: Student Reflections across Workshops 1 to 6
ref_summary = df_ref.pivot_table(
    index=['lab_number', 'session_date'],
    columns='institution_context',
    values='reflection_uid',
    aggfunc='count',
    fill_value=0
)
display(ref_summary)

ax = df_ref.groupby('lab_number')['reflection_uid'].count().plot(
    kind='bar', color='#2b5c8f', figsize=(8, 4),
    title='OPENSC 2026 - Anonymized Student Reflections by Workshop'
)
ax.set_xlabel('Workshop (Lab. 1–6)')
ax.set_ylabel('Number of Reflections')
plt.tight_layout()
plt.show()
```
